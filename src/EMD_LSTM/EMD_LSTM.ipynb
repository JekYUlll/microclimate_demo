# EMD + LSTM (Taishan)

This notebook runs an EMD-based decomposition and trains one LSTM per component, then
aggregates predictions for evaluation and plotting.


```python
from pathlib import Path
import sys

def find_project_root(start: Path) -> Path:
    markers = [{'src', 'data'}, {'src', 'scripts'}, {'src'}]
    for candidate in [start] + list(start.parents):
        present = {p.name for p in candidate.iterdir() if p.is_dir()}
        for marker in markers:
            if marker.issubset(present):
                return candidate
    return start

cwd = Path.cwd().resolve()
PROJECT_ROOT = find_project_root(cwd)
if (PROJECT_ROOT / 'src').exists() and str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
elif (cwd.parent / 'src').exists() and str(cwd.parent) not in sys.path:
    sys.path.insert(0, str(cwd.parent))

from src.EMD_LSTM.emd_lstm import ExperimentConfig, run_experiment
```

```python
cfg = ExperimentConfig(
    data_path=PROJECT_ROOT / 'data/AntAWS/3_hourly/Taishan_3h.csv',
    target_col='Temperature(Ąć)',
    feature_cols=None,
    encoding="latin1",          # 建议显式写编码
    input_window=24,
    horizon=6,
    train_ratio=0.8,
    epochs=30,
    batch_size=128,
    learning_rate=1e-3,
    hidden_size=128,
    num_layers=2,
    dropout=0.1,
    seed=42,
    device="cuda",
    max_imfs=2,                 # 限制 IMF 数量
    emd_max_sift=50,            # 每个 IMF 最大筛分次数，避免无限迭代
    emd_spline_kind="slinear",  # 简单插值，速度更快
    max_samples=20000,          
    max_points=2000,
    log_path=PROJECT_ROOT / 'log/emd_lstm_taishan.log',
    log_every=1,
    verbose=True,
    plot_path=PROJECT_ROOT / 'plots/emd_lstm_taishan.png',
    plot_last_days=5,
)

preds, targets, time_axis = run_experiment(cfg)
```
